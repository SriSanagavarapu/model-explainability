# Import

```python
from langchain_ollama import OllamaLLM
```

```python
client = OllamaLLM(model="llama3.2:latest", temperature=0)
```

```python
prompt = """I am trying to create a graph of AIML models and their explainability methods.
Explainability methods have also been expanded upon from research papers to include improvements on the original techniques and applications of the technique in creative ways.
Create a graph with the following types of nodes and relationships:

Nodes:
1. AIML Model Family (e.g., Tree-based Models, Neural Networks, etc.)
2. AIML Model Algorithm (e.g., Decision Trees, Random Forests, Neural Networks, etc.)
3. Explainability Method (e.g., SHAP, LIME, Grad-CAM, etc.)
4. Explainability Method Improvement (which include improvements in the inherent method e.g., Kernel SHAP, DeepLIFT, etc.)
5. Explainability Method Application (which include applications of the method in creative ways e.g., using SHAP for feature selection, etc.)

Relationships:
1. Explanation (This relationship connects AIML Model Algorithm or AIML Model Family to Explainability Method)
2. Utilization (This relationship connects Explainability Method to Explainability Method Improvement or Explainability Method Application)

Properties:
- AIML Model Family: name, description
- AIML Model Algorithm: name, description, year of introduction, initial research paper title
- Explainability Method: name, description, year of introduction, initial research paper title
- Explainability Method Improvement: name, description, year of introduction, research paper title
- Explainability Method Application: name, description, year of introduction, research paper title
- Explanation: type (e.g., post-hoc, intrinsic), purpose (what does the technique try to explain? training data, model predictions, algorithm itself, etc.), description
- Utilization: type (improvement or application), description of the improvement or application

First, describe each node individually with its properties. Start with 2-3 nodes for each type.
Then, describe each relationship individually with source and destination nodes with its properties."""
```

```python
response = client.invoke(prompt)
```

```python
from IPython.display import display, Markdown
display(Markdown(response))
```
