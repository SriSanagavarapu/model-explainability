```python
import networkx as nx
```

```python
G = nx.Graph()
G.add_node(1, prop="Hello", test=123)
```

```python
G.nodes(data=True)
```

```python
G.add_node(2, **{'prop': 'World', 'test': 456})
```

```python
G.add_nodes_from([(3, {'prop': 'Hello', 'test': 124}), (4)])
```

```python
G.nodes(data=True)
```

```python
G.add_edge(1, 2, connect=5)
```

```python
G.add_edges_from([(2, 3, {'connect':10}), (3, 4, {'connect':15})])
```

```python
print(G.nodes(data=True))
print(G.edges(data=True))
```

```python
G.add_edge(1, 4, weight=5)
```

```python
print(G.edges(data=True))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))
nx.draw(G, with_labels=True, node_color='lightblue', edge_color='gray', node_size=700)
plt.show()
```
